```python
#Importacion de las librerias necesarias
##Los comentarios añadidoS al iniciar la fase 3 se identificaran con un doble numeral (##)
import numpy as np
import pandas as pd
import time
import matplotlib.pyplot as plt
import seaborn as sns
#Funciones utilizadas en el codigo
#Funcion para saludar al usuario y funcion que nos mostrara datos porcentuales en la grafica generada mas adelante
def saludar(nombre_usuario):
    return f'Hola, {nombre_usuario}. ¡Espero que tengas un excelente día!'
def formato(pct):
    total = valores.sum()
    cantidad = int(round(pct * total / 100))
    return f'{pct:.2f}%\n({cantidad})'
#Leer el archivo CSV
Datos_restaurantes = pd.read_csv('base_datos_restaurantes_USA_v2.csv')
dr = Datos_restaurantes.copy()
ciudad_residencia = 'ciudad_residencia'
ciudades = dr[ciudad_residencia].dropna().unique()
#Ubicar y corregir de edades negativas o mayores a 120, reemplazarlas por un valor vacio y llenar esos valores vacios con la mediana con el fin de que no afecte a los datos
#Transformas el valor obtenido tipo float de la mediana a un valor de tipo entero
dr.loc[(dr['edad'] < 0) | (dr['edad'] > 120), 'edad'] = pd.NA
dr['edad'] = dr['edad'].fillna(dr['edad'].median())
dr['edad'] = dr['edad'].astype(int)
#Eliminar espacios en blanco al inicio y al final de los valores de texto para que no sean tomados como valores diferentes ni afecte a la eficacia del codigo
columnas_texto = [
    'nombre',
    'apellido',
    'genero',
    'ciudad_residencia',
    'estrato_socioeconomico',
    'ocio',
    'consume_licor',
    'preferencias_alimenticias',
    'membresia_premium',
    'telefono_contacto',
    'correo_electronico',
    'tipo_de_pago_mas_usado']
for col in columnas_texto:
    dr[col] = dr[col].str.strip()
#Ubicar y corregir valores negativos en 'frecuencia_visita' reemplezandolos por un espacio vacio y rellenandolos con la mediana
#Transformas el valor obtenido tipo float de la mediana a un valor de tipo entero
dr.loc[dr['frecuencia_visita'] < 0, 'frecuencia_visita'] = pd.NA
dr['frecuencia_visita'] = dr['frecuencia_visita'].fillna(dr['frecuencia_visita'].median())
dr['frecuencia_visita'] = dr['frecuencia_visita'].astype(int)
#Correcion de espacios en blanco rellenando los mismos con la mediana o con la cadena de texto mas adecuada
dr['promedio_gasto_comida'] = dr['promedio_gasto_comida'].fillna(dr['promedio_gasto_comida'].median())
dr['preferencias_alimenticias'] = (dr['preferencias_alimenticias'].fillna('Sin preferencia'))
dr['telefono_contacto'] = (dr['telefono_contacto'].fillna('No registrado'))
dr['correo_electronico'] = (dr['correo_electronico'].fillna('No registrado'))
#Reemplazo de caracteres que afectaban a la presentacion de los datos telefonicos
#Esto con el fin de mantener un orden claro y conciso para facilidad del usuario
dr['telefono_contacto'] = (
    dr['telefono_contacto']
    .str.replace('.', '-', regex=False)
    .str.replace('(', '', regex=False)
    .str.replace(')', '-', regex=False))
##Avance 3, aqui se mostraran las graficas solicitadas junto a su respectivo analisis
personas_ciudad = (
    dr['ciudad_residencia']
    .value_counts()
)
display(
    personas_ciudad
    .reset_index()
)
##Cantidad de personas por ciudad con el fin de identificar en que ciudad existe una mayor concentracion de clientes
print('\nAnalisis 1: Cantidad de personas por ciudad\n'
    'Este grafico permite identificar cuáles ciudades\n'
    'concentran la mayor cantidad de clientes dentro\n'
    'de la base de datos. Esta información resulta util\n'
    'para priorizar campañas de marketing y promociones')
plt.figure(figsize=(10,5))
personas_ciudad.plot(kind='bar')
plt.title('Cantidad de personas por ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad de personas')
plt.show()
##Distribución por estrato socioeconomico
estratos = (
    dr['estrato_socioeconomico']
    .value_counts()
)
display(
    estratos
    .reset_index()
)
print('\nAnalisis 2: Distribucion por estrato socioeconomico\n'
    'Este analisis permite conocer la composicion\n'
    'socioeconomica de los clientes y facilita la\n'
    'segmentacion de campañas comerciales segun\n'
    'el poder adquisitivo de cada grupo.')
ax = estratos.plot(kind='bar', figsize=(8,5))
total = estratos.sum()
for i, valor in enumerate(estratos):
    ax.text(i, valor, f'{valor/total*100:.1f}%',
            ha='center', va='bottom')
plt.title('Distribución por estrato socioeconómico')
plt.xlabel('Estrato')
plt.ylabel('Cantidad')
plt.show()
##Gasto promedio por ciudad
gasto_ciudad = (dr.groupby('ciudad_residencia')['promedio_gasto_comida'].mean().sort_values(ascending=False))
display(
    gasto_ciudad
    .reset_index()
)
print('\nAnalisis 3: Gasto promedio por ciudad\n'
    'Este grafico permite identificar las ciudades\n'
    'donde los clientes presentan mayores niveles\n'
    'de gasto, facilitando la toma de decisiones\n'
    'sobre promociones, alianzas y programas de fidelizacion.')
plt.figure(figsize=(10,5))
gasto_ciudad.plot(kind='bar')
plt.title('Gasto promedio por ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Gasto promedio')
plt.show()
##Frecuencia de visita vs gasto promedio según estrato
print('\nAnalisis 4: Frecuencia de visita y gasto promedio\n'
    'Este analisis permite observar si los clientes\n'
    'que visitan restaurantes con mayor frecuencia\n'
    'tambien realizan gastos mas altos y como este\n'
    'comportamiento varia entre estratos socioeconomicos.')
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=dr,
    x='frecuencia_visita',
    y='promedio_gasto_comida',
    hue='estrato_socioeconomico'
)
plt.title('Frecuencia de visita vs gasto promedio por estrato')
plt.show()
# Ingresos mensuales vs gasto promedio
print('\nAnalisis 5: Ingresos mensuales vs gasto promedio\n'
    'El grafico de dispersion permite identificar\n'
    'si existe una relacion entre los ingresos de\n'
    'los clientes y el dinero que destinan al consumo\n'
    'en restaurantes.')
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=dr,
    x='ingresos_mensuales',
    y='promedio_gasto_comida'
)
plt.title('Ingresos mensuales vs gasto promedio')
plt.show()
##Preferencias alimenticias por ciudad
preferencias = pd.crosstab(
    dr['ciudad_residencia'],
    dr['preferencias_alimenticias']
)
display(preferencias)
preferencias.plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)
print('\nAnalisis 6: Preferencias alimenticias por ciudad\n'
    'Este analisis permite identificar que tipos\n'
    'de alimentacion son mas populares entre los\n'
    'clientes y como dichas preferencias varian\n'
    'segun la ciudad de residencia.')
plt.title('Preferencias alimenticias por ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.show()
##Perfil de clientes de mayor gasto
umbral = (dr['promedio_gasto_comida'].quantile(0.90))
clientes_top = (dr[dr['promedio_gasto_comida']>= umbral])
print(f'\nClientes de mayor gasto analizados: '
    f'{len(clientes_top)}')
print('\nAnalisis 7: Perfil de clientes de mayor gasto\n'
    'Se analizan los clientes que presentan los\n'
    'niveles de gasto mas altos con el fin de\n'
    'identificar patrones relacionados con edad,\n'
    'preferencias alimenticias, ingresos y ubicacion.')
display(
    clientes_top[
        [
            'edad',
            'ciudad_residencia',
            'preferencias_alimenticias',
            'promedio_gasto_comida',
            'ingresos_mensuales'
        ]
    ].head(20)
)
print('\nPreferencias alimenticias más frecuentes entre los clientes de mayor gasto')
display(
    clientes_top[
    'preferencias_alimenticias'
    ].value_counts()
)
##Ciudad con más membresías premium
premium = (dr[dr['membresia_premium']== 'Sí']['ciudad_residencia'].value_counts())
print('\nCantidad de membresías premium por ciudad')
display(
    premium
    .reset_index()
)
print('\nAnalisis 8: Membresias premium por ciudad\n'
    'Este analisis permite determinar en que ciudades\n'
    'los clientes muestran una mayor aceptacion de\n'
    'programas premium y servicios de fidelizacion.')
plt.figure(figsize=(10,5))
premium.plot(kind='bar')
plt.title('Membresías premium por ciudad')
plt.xlabel('Ciudad')
plt.ylabel('Cantidad')
plt.show()
##Relación entre consumo de alcohol y edad
print('\nAnalisis 9: Consumo de alcohol y edad\n'
    'Este grafico permite comparar la distribucion\n'
    'de edades entre los clientes que consumen\n'
    'alcohol en restaurantes y aquellos que no lo hacen.')
plt.figure(figsize=(8,6))
sns.boxplot(
    data=dr,
    x='consume_licor',
    y='edad'
)
plt.title('Consumo de alcohol según edad')
plt.show()
##INTEGRACIÓN DE DATOS YELP
yelp = pd.read_csv('yelp_restaurants.csv')
print(f'\nRestaurantes únicos encontrados en Yelp: '
    f'{yelp['id'].nunique()}')
##Categorías más frecuentes en Yelp
categorias_yelp = (
    yelp['title']
    .value_counts()
    .head(10)
)
print('\nTop 10 categorías más frecuentes en Yelp')
display(
    categorias_yelp
    .reset_index()
)
print('\nAnalisis 10: Integracion de datos yelp\n'
    'Se incorporan datos provenientes de yelp para\n'
    'enriquecer el analisis de clientes mediante\n'
    'informacion sobre categorias de restaurantes,\n'
    'calificaciones, popularidad y tendencias del mercado.')
plt.figure(figsize=(10,5))
categorias_yelp.plot(kind='bar')
plt.title('Top 10 categorías más frecuentes en Yelp')
plt.xlabel('Categoría')
plt.ylabel('Cantidad')
plt.show()
##Restaurantes mejor calificados
top_rating = (
    yelp[
        [
            'name',
            'title',
            'rating',
            'review_count'
        ]
    ]
    .sort_values(
        by='rating',
        ascending=False
    )
    .head(10)
)
print('\nRestaurantes mejor calificados')
display(top_rating)
##Restaurantes más populares
top_reviews = (
    yelp[
        [
            'name',
            'title',
            'rating',
            'review_count'
        ]
    ]
    .sort_values(
        by='review_count',
        ascending=False
    )
    .head(10)
)
print('\nRestaurantes con más reseñas')
display(top_reviews)
##Estadísticas generales Yelp
estadisticas_yelp = pd.DataFrame({
    'Indicador': [
        'Restaurantes únicos',
        'Calificación promedio',
        'Calificación máxima',
        'Calificación mínima',
        'Promedio de reseñas'
    ],
    'Valor': [
        yelp['id'].nunique(),
        round(yelp['rating'].mean(),2),
        yelp['rating'].max(),
        yelp['rating'].min(),
        round(yelp['review_count'].mean(),0)
    ]
})
print('\nEstadísticas generales Yelp')
display(estadisticas_yelp)
# Conclusiones rápidas
print('\nSe te envio un documento por separado con ciertas recomendaciones, revisalo cuando estes listo\n')
#Invitacion al usuario y saludo
nombre_usuario = input('Escriba su usuario para continuar: ')
print(saludar(nombre_usuario))
time.sleep(2)
print('En breve tendras acceso a la base de datos de restaurantes en USA\nPara su facil acceso puedes acceder a los datos filtrados por ciudad')
time.sleep(4)
print('Aqui tienes una grafica de distribucion de datos por ciudad\nCierrala cuando estes listo para continuar')
time.sleep(2)
ciudades = {
    1: 'Miami',
    2: 'Denver',
    3: 'Boston',
    4: 'San Diego',
    5: 'Dallas',
    6: 'NYC',
    7: 'Seattle',
    8: 'Chicago',
    9: 'Houston',
    10: 'Phoenix'
}
#Impresion de gafica tipo pie con la intencion de que el usuario pueda ver la cantidad de datos total y sus porcentajes con la finalidad de que ayude a la toma de decisiones del usuario conforme a cuantos datos le gustaria analizar
orden_ciudades = list(ciudades.values())
valores = dr['ciudad_residencia'].value_counts().reindex(orden_ciudades, fill_value=0)
etiquetas = [f'{numero}. {ciudad}'for numero, ciudad in ciudades.items()]
plt.figure(figsize=(10, 8))
plt.pie(valores, labels=etiquetas, autopct=formato)
plt.title(f'Distribución de datos por ciudad ({len(dr)})')
plt.show()
#Bucle while de apertura con la intension de que al entrar al siguiente ciclo while y salir del mismo el codigo no termine, sino que el usuario tenga la posibilidad de volver a explorar diferentes opciones
#Esto con la finalidad de que el usuario pueda explorar los datos de las ciudades por separado y pueda cambiar de ciudad a su comodidad
#El usuario se movera atraves del codigo ingresando los digitos disponibles, cualquier valor fuera de rango le solicitara al usuario que trate nuevamente
#Incluso con valores erroneos el codigo no se interrumpira sino que insistira en recibir una de las respuestas validas, para que el codigo no se rompa mientras el usuario esta usandolo
while True:
    for numero, ciudad in ciudades.items():
        print(f'{numero}. {ciudad}')
    print('0. Salir')
    try:
        opcion = int(input('Seleccione una ciudad: '))
        if opcion == 0:
            print(f'Hasta luego, {nombre_usuario}!')
            break
        if opcion not in ciudades:
            print('Ciudad inválida, intente nuevamente.')
            continue
        #Creacion de las variables necesarios para seguir analizando los datos solicitados por el usuario asi como variables para el control del codigo y su correcta funcionalidad
        ciudad_seleccionada = ciudades[opcion]
        datos_filtrados = dr[dr['ciudad_residencia'] == ciudad_seleccionada]
        max_filas = len(datos_filtrados)
        print(f'\nCiudad: {ciudad_seleccionada}')
        time.sleep(1)
        print(f'Registros disponibles: 0 a {max_filas}')
        salir_filas = False
        #Menú de filas
        #Segundo ciclo while para que el usuario pueda explorar filas de datos
        #Contencion de posibles errores de digitacion del usuario o ingreso de valores no validos
        while True:
            try:
                time.sleep(1)
                inicio = int(input('Seleccione la fila inicial: '))
                fin = int(input('Seleccione la fila final: '))
                if not (0 <= inicio < fin <= max_filas):
                    print(f'Error: seleccione un rango entre 0 y {max_filas}.\nTenga en cuenta que la fila inicial debe ser menor que la final.')
                    continue
                if 0 <= inicio < fin <= max_filas:
                    print(datos_filtrados.iloc[inicio:fin])
                #Tercer cliclo while para que el usuario pueda seguir explorando diferentes filas de datos si asi lo desea o volver al menu anterior de ciudades
                #Nuevamente control de datos invalidos para el codigo sin detener su funcion
                while True:
                    time.sleep(2)
                    accion = input(
                    '\n1. Ver otras filas\n'
                    '2. Volver al menú de ciudades\n'
                    'Seleccione una opción: ')
                    if accion == '1':
                        break
                    elif accion == '2':
                        print('Volviendo al menú de ciudades...')
                        salir_filas = True
                        break
                    else:
                        print('Opción inválida, intente nuevamente.')
                if salir_filas:
                    break
            except ValueError:
                print('Debes ingresar números.')
    except ValueError:
        print('Debes ingresar números.')
```
